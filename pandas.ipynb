```python
import pandas as pd

students_data = {
    "Name": ["Aisha", "Rahul", "Sara", "John", "Meera"],
    "Age": [21, 22, 20, 23, 21],
    "Python": [85, 78, 95, 65, 88],
    "Statistics": [90, 82, 91, 72, 86],
    "MachineLearning": [88, 80, 94, 68, 90]
}

df = pd.DataFrame(students_data)

df.to_csv("students.csv", index=False)

print("students.csv created successfully!")
print(df)
```

```python
print("\nDataset shape:", df.shape)
print("\nColumn information:",df.info())
```

```python
subjects = ["Python", "Statistics", "MachineLearning"]
df["Average"] = df[subjects].mean(axis=1).round(2)
print(df[["Name", "Average"]])
```

```python
high_performers = df[df["Average"] >= 75]
print(high_performers[["Name", "Average"]])
```

```python
top_student = df.loc[df["Average"].idxmax()]

print("Top student:", top_student["Name"])
print("Average marks:", top_student["Average"])
```

```python
high_performers.to_csv("high_performers.csv", index=False)

print("High performers saved successfully!")
```
